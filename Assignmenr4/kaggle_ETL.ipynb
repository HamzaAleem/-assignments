```python
import pandas as pd
df = pd.read_csv('retail-orders/orders.csv',na_values = ['Not Available','unknown'])

df.head(20)
```

```python
df.columns=df.columns.str.replace(' ','-')
```

```python
df.head()
df['Discount-Price']=df['List-Price']*df['Discount-Percent']*.01
df['Sell-Price'] = df['List-Price']-df['Discount-Price']
df['Sales']=df['Sell-Price']*df['Quantity']
df.head(20)
```
